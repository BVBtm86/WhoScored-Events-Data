### Help

```python
class ReportError:
    def __init__(self, config: ScrapeDataConfig):
        self.config = config

        self.client = MongoClient(config.mongo.url)
        self.db = self.client[config.mongo.db]

        logs_collection_name = config.mongo.collection["collection_logs"]
        self.collection = self.db[logs_collection_name]

    def report(self, job_name: str, exc: Exception, context: Optional[Dict[str, Any]] = None) -> None:
        context = context or {}

        error_doc = {
            "job": job_name,
            "timestamp": datetime.now(timezone.utc),
            "host": socket.gethostname(),
            "season_year": self.config.season.year,
            "league": self.config.season.league,
            "competition": self.config.season.name,
            "country": self.config.season.country,
            "context": context,
            "error_type": type(exc).__name__,
            "error_message": str(exc),
            "traceback": traceback.format_exc(),
        }

        # Log to Mongo
        try:
            self.collection.insert_one(error_doc)
        except Exception:
            pass

        # Email alert (only if SMTP password exists)
        if self.config.email.password:
            self._send_email(error_doc)

    def _send_email(self, error_doc: dict) -> None:
        cfg = self.config.email

        subject = f"[WhoScored Scrape ERROR] {error_doc['job']} | {self.config.season.name} {self.config.season.year}"
        body = (
            f"Job: {error_doc['job']}\n"
            f"Time (UTC): {error_doc['timestamp']}\n"
            f"Host: {error_doc['host']}\n\n"
            f"League/Season: {self.config.season.league} / {self.config.season.year}\n"
            f"Competition: {self.config.season.name} ({self.config.season.country})\n\n"
            f"Context: {error_doc['context']}\n\n"
            f"{error_doc['error_type']}: {error_doc['error_message']}\n\n"
            f"Traceback:\n{error_doc['traceback']}\n"
        )

        msg = EmailMessage()
        msg["Subject"] = subject
        msg["From"] = cfg.from_email
        msg["To"] = cfg.to_email
        msg.set_content(body)

        try:
            if cfg.use_tls:
                with smtplib.SMTP(cfg.smtp_host, cfg.smtp_port, timeout=20) as server:
                    server.starttls()
                    server.login(cfg.username, cfg.password)
                    server.send_message(msg)
            else:
                with smtplib.SMTP_SSL(cfg.smtp_host, cfg.smtp_port, timeout=20) as server:
                    server.login(cfg.username, cfg.password)
                    server.send_message(msg)
        except Exception:
            pass
```

### Game Schedule

```python
class GameSchedule:
    FINISHED_STATUS_CODE = 6

    def __init__(self, config: ScrapeDataConfig, reporter: ReportError):
        self.config = config
        self.reporter = reporter

        self.ws = sd.WhoScored(leagues=config.season.league, 
                               seasons=config.season.year)

        self.client = MongoClient(config.mongo.url)
        self.db = self.client[config.mongo.db]

        schedule_collection_name = config.mongo.collection["collection_schedule"]
        self.collection_schedule = self.db[schedule_collection_name]

    def _read_season_schedule(self) -> pd.DataFrame:
        return self.ws.read_schedule().reset_index(drop=False)

    def _extract_finished_game_ids(self) -> Set[int]:
        cursor = self.collection_schedule.find(
            {"game_status": "finished"},
            {"_id": 0, "game_id": 1},
        )
        return {int(doc["game_id"]) for doc in cursor if doc.get("game_id") is not None}

    def _get_newly_finished_games(self) -> pd.DataFrame:

        df = self._read_season_schedule()

        required = {"game_id", "status", "start_time"}
        missing = required - set(df.columns)
        if missing:
            raise ValueError(f"WhoScored schedule missing columns: {sorted(missing)}")

        finished_ids = self._extract_finished_game_ids()

        df_remaining = df[~df["game_id"].astype(int).isin(finished_ids)].copy()
        df_new_finished = df_remaining[df_remaining["status"] == self.FINISHED_STATUS_CODE].copy()

        if df_new_finished.empty:
            return pd.DataFrame()

        return df_new_finished[["game_id", "start_time"]].reset_index(drop=True)

    def save_schedule(self) -> None:
        job_name = "GameSchedule.save_schedule"

        try:
            df_new_finished = self._get_newly_finished_games()

            if df_new_finished.empty:
                print("No new finished games found.")
                return

            ops = []
            now = datetime.now(timezone.utc)

            for _, row in df_new_finished.iterrows():
                game_id = int(row["game_id"])
                start_time = row.get("start_time")

                doc = {
                    "game_id": game_id,
                    "game_date": start_time,
                    "season": self.config.season.year,
                    "competition_name": self.config.season.name,
                    "competition_country": self.config.season.country,
                    "game_status": "finished",
                }

                ops.append(
                    UpdateOne(
                        {"game_id": game_id},
                        {"$set": doc, "$setOnInsert": {"created_at": now}},
                        upsert=True,
                    )
                )

            result = self.collection_schedule.bulk_write(ops, ordered=False)

            affected = result.upserted_count + result.modified_count

            logger.info(f"Finished games processed: {len(df_new_finished)}")
            logger.info(f"Mongo affected documents: {affected}")

        except Exception as e:
            self.reporter.report(
                job_name=job_name,
                exc=e,
                context={
                    "season_year": self.config.season.year,
                    "league": self.config.season.league,
                },
            )
            raise
```

### Extract Raw New Games

```python
class RawEvents:

    def __init__(self, config: ScrapeDataConfig, reporter: ReportError):
        self.config = config
        self.reporter = reporter

        self.client = MongoClient(config.mongo.url)
        self.db = self.client[config.mongo.db]

        schedule_name = config.mongo.collection["collection_schedule"]
        self.collection_schedule = self.db[schedule_name]

        raw_events_name = config.mongo.collection.get("collection_raw_events")
        self.collection_raw_events = self.db[raw_events_name]

    def _finished_games_df(self) -> pd.DataFrame:
 
        docs = list(self.collection_schedule.find(
            {"season": scrape_config.season.year,
            "game_status": "finished"},
            {"_id": 0}
        ))
        return pd.DataFrame(docs)

    def _raw_events_extracted(self) -> Set[int]:

        return set(self.collection_raw_events.distinct("game_id"))

    def _process_raw_events(self, 
                           game_info: pd.DataFrame, 
                           events: pd.DataFrame) -> List[dict]:

        events = events.reset_index(drop=False)
        for col in ["league", "season", "game"]:
            if col in events.columns:
                events = events.drop(columns=[col])

        info = game_info.copy()
        merged = pd.merge(info, events, on="game_id", how="right")

        return merged.to_dict(orient="records")

    def save_new_finished_games(self) -> None:
        
        job_name = "RawEvents.save_new_finished_games"

        try:
            games_df = self._finished_games_df()

            if games_df.empty:
                logger.info("No finished games found in schedule collection.")
                return

            extracted_game_ids = self._raw_events_extracted()

            game_ids = [int(x) for x in games_df["game_id"].tolist()]
            to_process = [gid for gid in game_ids if gid not in extracted_game_ids]

            if not to_process:
                logger.info("No new games to process (raw events already present).")
                return

            logger.info(
                "Games to process: %d (skipping %d already saved)",
                len(to_process),
                len(game_ids) - len(to_process),
            )

            games_info = games_df[games_df["game_id"].isin(to_process)].reset_index(drop=True)
            ws = sd.WhoScored(
                leagues=self.config.season.league,
                seasons=scrape_config.season.year,
            )

            inserted_games = 0
            inserted_rows_total = 0

            for game_id in to_process:
                game_id = int(game_id)

                try:

                    events = ws.read_events(match_id=game_id, output_fmt="events")

                    if events is None or len(events) == 0:
                        logger.info("No events for game_id=%s (skipping).", game_id)
                        continue

                    game_info = games_info[games_info["game_id"] == game_id].reset_index(drop=True)
                    records = self._process_raw_events(game_info=game_info, events=events)

                    if not records:
                        logger.info("No merged records for game_id=%s (skipping).", game_id)
                        continue

                    self.collection_raw_events.insert_many(records, ordered=False)

                    inserted_games += 1
                    inserted_rows_total += len(records)

                    logger.info("Inserted game_id=%s | rows=%d", game_id, len(records))

                except Exception as e:
                    self.reporter.report(
                        job_name="RawEvents.insert_game",
                        exc=e,
                        context={
                            "game_id": game_id,
                            "league": self.config.season.league,
                        },
                    )
                    logger.exception("Error processing game_id=%s", game_id)

            logger.info("Process Raw Events complete. Games inserted: %d | Total rows inserted: %d",
                        inserted_games, inserted_rows_total)

        except Exception as e:
            self.reporter.report(
                job_name=job_name,
                exc=e,
                context={"season_year": self.config.season.year, "league": self.config.season.league},
            )
            raise
```

```python
from datetime import datetime
from pathlib import Path
import json
import pandas as pd


def _append_audit_record(output_path: str, game_id, issues: list[dict]) -> None:
    record = {
        "game_id": game_id,
        "timestamp": datetime.utcnow().isoformat(),
        "issues": issues,
    }
    path = Path(output_path)
    with path.open("a", encoding="utf-8") as f:
        f.write(json.dumps(record) + "\n")


def audit_single_game_shots(
    events_df: pd.DataFrame,
    game_id,
    output_path: str = "shot_audit_issues.jsonl",
    team_col: str = "team",
) -> bool:
    """
    Runs validation on ONE game dataframe.

    Appends to file ONLY if problems found.
    Returns True if issue detected, else False.
    """
    df = events_df.copy()
    issues: list[dict] = []

    if "is_shot" not in df.columns:
        return False

    shot_df = df[df["is_shot"].fillna(0).astype(int) == 1].copy()

    # -------------------------------------------------
    # 0) No shots detected but shot events exist
    # -------------------------------------------------
    if shot_df.empty:
        if "type" in df.columns and df["type"].isin({"Goal", "SavedShot", "MissedShots", "ShotOnPost", "OwnGoal"}).any():
            issues.append({"type": "shots_missing_detection"})
        if not issues:
            return False
        _append_audit_record(output_path, game_id, issues)
        return True

    # Ensure team column exists
    if team_col not in shot_df.columns:
        shot_df[team_col] = "ALL"
    else:
        shot_df[team_col] = shot_df[team_col].fillna("UNKNOWN_TEAM")

    # -------------------------------------------------
    # 1) Unknown zones
    # -------------------------------------------------
    if "Shot_Zone_Unknown" in shot_df.columns:
        unk = shot_df[shot_df["Shot_Zone_Unknown"].fillna(False).astype(bool)]
        if not unk.empty:
            issues.append({"type": "unknown_zone", "count": int(len(unk))})

    # # Unknown situation (optional but useful)
    # if "Shot_Sit_Unknown" in shot_df.columns:
    #     unk = shot_df[shot_df["Shot_Sit_Unknown"].fillna(False).astype(bool)]
    #     if not unk.empty:
    #         issues.append({"type": "unknown_situation", "count": int(len(unk))})

    # -------------------------------------------------
    # 2) Logical conflicts (row-level)
    # -------------------------------------------------
    def _b(s: pd.Series) -> pd.Series:
        return s.fillna(False).astype(bool)

    if {"Shot_OnTarget", "Shot_Blocked"}.issubset(shot_df.columns):
        both = shot_df[_b(shot_df["Shot_OnTarget"]) & _b(shot_df["Shot_Blocked"])]
        if not both.empty:
            issues.append({"type": "on_target_and_blocked", "count": int(len(both))})

    if {"Shot_OnTarget", "Shot_OffTarget"}.issubset(shot_df.columns):
        both = shot_df[_b(shot_df["Shot_OnTarget"]) & _b(shot_df["Shot_OffTarget"])]
        if not both.empty:
            issues.append({"type": "on_target_and_off_target", "count": int(len(both))})

    # -------------------------------------------------
    # 3) Per-team sum checks
    # -------------------------------------------------
    sum_zones = ["Shot_Zone_6Yard", "Shot_Zone_Box", "Shot_Zone_OutBox", "Shot_Zone_Unknown"]
    sum_situation = [
        "Shot_Sit_Penalty",
        "Shot_Sit_Fastbreak",
        "Shot_Sit_SetPieces",
        "Shot_Sit_OpenPlay",
        "Shot_Sit_Unknown",
        "Shot_Sit_OwnGoal",
    ]
    # Body parts: your pipeline is Head/Right/Left/Other (no Unknown)
    sum_parts = ["Shot_BP_Head", "Shot_BP_Right", "Shot_BP_Left", "Shot_BP_Other"]

    # Result buckets (should be mutually exclusive on non-OG, non-goal shots)
    sum_results = ["Shot_Woodwork", "Shot_Blocked", "Shot_OnTarget", "Shot_OffTarget"]

    def _safe_sum_bool(g: pd.DataFrame, cols: list[str]) -> int:
        cols = [c for c in cols if c in g.columns]
        if not cols:
            return 0
        return int(g[cols].fillna(False).astype(bool).to_numpy().sum())

    def _col_sums(g: pd.DataFrame, cols: list[str]) -> dict:
        out = {}
        for c in cols:
            if c in g.columns:
                out[c] = int(g[c].fillna(False).astype(bool).sum())
        return out

    for team, g in shot_df.groupby(team_col, dropna=False):
        total_shots = int(len(g))
        own_goals = int(_b(g["Shot_Own"]).sum()) if "Shot_Own" in g.columns else 0
        goals = int(_b(g["Shot_Goal"]).sum()) if "Shot_Goal" in g.columns else 0

        # ---- ZONES / SITUATION / BODY PARTS ----
        # With your convention, zones/situation/body parts are zeroed for OG,
        # so expected sum = total_shots - own_goals.
        expected_non_og = total_shots - own_goals

        if all(c in g.columns for c in sum_zones):
            zones_sum = _safe_sum_bool(g, sum_zones)
            if zones_sum != expected_non_og:
                issues.append({
                    "type": "shot_zones_mismatch",
                    "team": team,
                    "total_shots": total_shots,
                    "own_goals": own_goals,
                    "expected": expected_non_og,
                    "actual": zones_sum,
                    "by_zone": _col_sums(g, sum_zones),
                })

        if all(c in g.columns for c in sum_situation):
            sit_sum = _safe_sum_bool(g, sum_situation)
            # includes Shot_Sit_OwnGoal, so expected should be total_shots (not non-OG)
            # BUT you zero situations for OG in add_shots, so Shot_Sit_OwnGoal is False for OG too.
            # Therefore expected is still expected_non_og.
            if sit_sum != expected_non_og:
                issues.append({
                    "type": "shot_situation_mismatch",
                    "team": team,
                    "total_shots": total_shots,
                    "own_goals": own_goals,
                    "expected": expected_non_og,
                    "actual": sit_sum,
                    "by_situation": _col_sums(g, sum_situation),
                })

        if all(c in g.columns for c in sum_parts):
            parts_sum = _safe_sum_bool(g, sum_parts)
            if parts_sum != expected_non_og:
                issues.append({
                    "type": "shot_parts_mismatch",
                    "team": team,
                    "total_shots": total_shots,
                    "own_goals": own_goals,
                    "expected": expected_non_og,
                    "actual": parts_sum,
                    "by_bodypart": _col_sums(g, sum_parts),
                })

        # ---- RESULTS ----
        # What you asked: (Woodwork + Blocked + OnTarget + OffTarget) should match
        # total_shots - goals - own_goals
        # Because:
        # - total_shots includes goals
        # - goals are included in Shot_OnTarget
        # - own goals should be excluded from buckets
        if all(c in g.columns for c in sum_results):
            is_og = _b(g["Shot_Own"]) if "Shot_Own" in g.columns else pd.Series(False, index=g.index)
            is_goal = _b(g["Shot_Goal"]) if "Shot_Goal" in g.columns else pd.Series(False, index=g.index)

            res = g[sum_results].fillna(False).astype(bool)
            res_row_sum = res.sum(axis=1)

            # For OG: expect 0 buckets (because you zero them)
            bad_og = g[is_og & (res_row_sum != 0)]
            if not bad_og.empty:
                issues.append({
                    "type": "own_goal_results_not_zeroed",
                    "team": team,
                    "count": int(len(bad_og)),
                    "expected_per_row": 0,
                })

            # For goals (non-OG): we EXPECT Shot_OnTarget=True, but for "exclude-goals sum"
            # goals should be ignored in the total bucket sum check below.
            # Row-level exclusivity check for NON-OG + NON-GOAL shots: exactly 1 bucket.
            bad_non_goal = g[~is_og & ~is_goal & (res_row_sum != 1)]
            if not bad_non_goal.empty:
                issues.append({
                    "type": "shot_results_not_exactly_one_bucket_non_goal",
                    "team": team,
                    "count": int(len(bad_non_goal)),
                    "expected_per_row": 1,
                })

            # Team-level bucket sum excluding goals and OG
            expected_excl_goals = total_shots - goals - own_goals

            # sum buckets on non-OG, non-goal only
            res_sum_excl = int(res.loc[~is_og & ~is_goal].to_numpy().sum())

            if res_sum_excl != expected_excl_goals:
                issues.append({
                    "type": "shot_results_sum_excluding_goals_mismatch",
                    "team": team,
                    "total_shots": total_shots,
                    "goals": goals,
                    "own_goals": own_goals,
                    "expected": expected_excl_goals,
                    "actual": res_sum_excl,
                    "by_result": {
                        "Shot_Woodwork": int(_b(g.loc[~is_og & ~is_goal, "Shot_Woodwork"]).sum()) if "Shot_Woodwork" in g.columns else 0,
                        "Shot_Blocked": int(_b(g.loc[~is_og & ~is_goal, "Shot_Blocked"]).sum()) if "Shot_Blocked" in g.columns else 0,
                        "Shot_OnTarget": int(_b(g.loc[~is_og & ~is_goal, "Shot_OnTarget"]).sum()) if "Shot_OnTarget" in g.columns else 0,
                        "Shot_OffTarget": int(_b(g.loc[~is_og & ~is_goal, "Shot_OffTarget"]).sum()) if "Shot_OffTarget" in g.columns else 0,
                    },
                })

    if not issues:
        return False

    _append_audit_record(output_path, game_id, issues)
    return True
```
